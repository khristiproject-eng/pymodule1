```python
import torch
import os
import psutil

process = psutil.Process(os.getpid())

# -------- BEFORE --------
print("==== BEFORE ====")
print("CPU Memory:", process.memory_info().rss / 1024**2, "MB")

if torch.cuda.is_available():
    print("GPU Memory:", torch.cuda.memory_allocated() / 1024**2, "MB")
else:
    print("GPU not available")


# -------- AFTER CPU TENSOR --------
a_cpu = torch.randn(5000, 5000)

print("\n==== AFTER CPU TENSOR ====")
print("CPU Memory:", process.memory_info().rss / 1024**2, "MB")

if torch.cuda.is_available():
    print("GPU Memory:", torch.cuda.memory_allocated() / 1024**2, "MB")


# -------- AFTER GPU TENSOR --------
if torch.cuda.is_available():
    a_gpu = torch.randn(5000, 5000).to("cuda")

    print("\n==== AFTER GPU TENSOR ====")
    print("CPU Memory:", process.memory_info().rss / 1024**2, "MB")
    print("GPU Memory:", torch.cuda.memory_allocated() / 1024**2, "MB")
```

```python
import torch
import time

size = 4000  # increase if needed

# -------- CPU --------
a = torch.randn(size, size)
b = torch.randn(size, size)

start = time.time()
c = a @ b
end = time.time()

print("CPU Time:", end - start, "seconds")


# -------- GPU --------
if torch.cuda.is_available():
    a = a.to("cuda")
    b = b.to("cuda")

    torch.cuda.synchronize()  # important for correct timing
    start = time.time()

    c = a @ b

    torch.cuda.synchronize()  # wait for GPU to finish
    end = time.time()

    print("GPU Time:", end - start, "seconds")
else:
    print("CUDA not available")
```

```python
import torch
import time

# Fake dataset
X = torch.randn(10000, 100)
y = torch.randn(10000, 1)

# Simple model (like a tiny neural network)
model = torch.nn.Linear(100, 1)

# Loss + optimizer
loss_fn = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# -------- CPU TRAINING --------
start = time.time()

for i in range(200):
    pred = model(X)
    loss = loss_fn(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

end = time.time()
print("CPU Training Time:", end - start, "seconds")


# -------- GPU TRAINING --------
if torch.cuda.is_available():

    device = "cuda"

    X = X.to(device)
    y = y.to(device)
    model = torch.nn.Linear(100, 1).to(device)

    loss_fn = torch.nn.MSELoss()
    optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

    torch.cuda.synchronize()
    start = time.time()

    for i in range(200):
        pred = model(X)
        loss = loss_fn(pred, y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    torch.cuda.synchronize()
    end = time.time()

    print("GPU Training Time:", end - start, "seconds")
else:
    print("CUDA not available")
```
